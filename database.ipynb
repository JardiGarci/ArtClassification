```python
import sqlite3

con = sqlite3.connect("data.db")
con.execute("PRAGMA journal_mode=WAL")
con.execute("PRAGMA foreign_keys=ON")


# Create metadata table: one row per painting with its WikiArt labels (artist, genre, movement).
# painting_id matches the image index in the Hugging Face dataset.

con.execute("""
CREATE TABLE IF NOT EXISTS metadata (
    painting_id  INTEGER PRIMARY KEY,
    artist       TEXT NOT NULL,
    genre        TEXT NOT NULL,
    movement     TEXT NOT NULL
)
""")

con.commit()
con.close()
```
